### Model Integration with Open AI, Gemini and Groq

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("openai/gpt-oss-120b", model_provider="groq")
model
```

```python
response = model.invoke("Hi, why do parrots talk?")
response
```

## Streaming

```python
for chunk in model.stream("Give me some random fact about penguins"):
    print(chunk.text, end="", flush=True)
```

## Batching

```python
questions = ["Who let the dogs out?", "Why is melody so chocolatey?", "Why are penguins black and white?"]

response_list = model.batch(questions)

for response in response_list:
    print(response.text)
```

### Batching with concurrency

Max concurrency tells it how many parallel calls can it make to the LLM for processing

```python
questions = ["Who let the dogs out?", "Why is melody so chocolatey?", "Why are penguins black and white?"]

response_list = model.batch(questions, config={"max_concurrency": 2}) # Only 2 requests will be sent at a time

for response in response_list:
    print(response.text)
```
